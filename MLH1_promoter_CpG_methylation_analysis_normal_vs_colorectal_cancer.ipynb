```python
sequence = "ATGCGTACGATCGATCGATCGTAGCTAGCTAGCTAGCTAGCGATCGATCGATCGTAGCTA"
print(f"Sequence: {sequence}")
print(f"Length: {len(sequence)} bp")
```

```python
g = sequence.count('G')
c = sequence.count('C')
gc = round(((g + c) / len(sequence)) * 100, 2)

print(f"GC Content: {gc}%")

if gc > 60:
    print("High GC — associated with CpG islands and gene promoters")
elif gc > 40:
    print("Moderate GC content")
else:
    print("Low GC content")
```

```python
import matplotlib.pyplot as plt

counts = {
    'A': sequence.count('A'),
    'T': sequence.count('T'),
    'G': sequence.count('G'),
    'C': sequence.count('C')
}

print("Nucleotide Counts:", counts)

plt.figure(figsize=(6,4))
plt.bar(counts.keys(), counts.values(),
        color=['#2ecc71','#e74c3c','#3498db','#f39c12'])
plt.title('Nucleotide Composition')
plt.xlabel('Nucleotide')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
mrna = sequence.replace('T', 'U')
print(f"mRNA: {mrna}")
```

```python
motif = "CGAT"
positions = []
start = 0

while True:
    pos = sequence.find(motif, start)
    if pos == -1:
        break
    positions.append(pos + 1)
    start = pos + 1

if positions:
    print(f"Motif '{motif}' found at positions: {positions}")
else:
    print(f"Motif '{motif}' not found")
```

```python
a = sequence.count('A')
t = sequence.count('T')
g = sequence.count('G')
c = sequence.count('C')
gc = round(((g + c) / len(sequence)) * 100, 2)

print(f"A: {a} | T: {t} | G: {g} | C: {c}")
print(f"GC Content: {gc}%")

if gc > 55:
    print("High GC content — consistent with a CpG island promoter region")
elif gc > 40:
    print("Moderate GC content")
else:
    print("Low GC content — may not be a CpG island")
```

```python
def find_cpg_positions(seq):
    positions = []
    for i in range(len(seq) - 1):
        if seq[i] == 'C' and seq[i+1] == 'G':
            positions.append(i + 1)  # 1-based position
    return positions

cpg_positions = find_cpg_positions(sequence)

print(f"Total CpG sites found: {len(cpg_positions)}")
print(f"CpG positions: {cpg_positions}")

# CpG observed vs expected ratio
# This tells you if this is a true CpG island
observed_cpg = len(cpg_positions)
expected_cpg = (sequence.count('C') * sequence.count('G')) / len(sequence)
ratio = round(observed_cpg / expected_cpg, 2)

print(f"\nCpG Observed/Expected ratio: {ratio}")
if ratio > 0.6:
    print("Ratio > 0.6 — this qualifies as a CpG island")
else:
    print("Ratio < 0.6 — not a classic CpG island")
```

```python
total_cpgs = len(cpg_positions)

# In normal tissue — none methylated
normal_methylated = []
normal_unmethylated = cpg_positions.copy()

# In cancer — majority methylated (literature documents ~70-80% methylation)
cancer_methylated = cpg_positions[:int(total_cpgs * 0.75)]
cancer_unmethylated = cpg_positions[int(total_cpgs * 0.75):]

print(f"Normal tissue — Methylated CpGs: {len(normal_methylated)}")
print(f"Normal tissue — Unmethylated CpGs: {len(normal_unmethylated)}")
print(f"\nCancer tissue — Methylated CpGs: {len(cancer_methylated)}")
print(f"Cancer tissue — Unmethylated CpGs: {len(cancer_unmethylated)}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 6))

# --- Normal tissue plot ---
for pos in normal_unmethylated:
    ax1.plot([pos, pos], [0, 1], color='gray', linewidth=0.8)
    ax1.scatter(pos, 1, s=80, facecolors='white',
                edgecolors='black', linewidth=1.5, zorder=3)

ax1.axhline(y=0, color='black', linewidth=2)
ax1.set_xlim(0, len(sequence))
ax1.set_ylim(-0.5, 1.8)
ax1.set_yticks([])
ax1.set_title('Normal Colorectal Tissue — MLH1 Promoter CpG Methylation',
              fontsize=12, fontweight='bold')
ax1.set_xlabel('Position (bp)')
ax1.text(10, 1.4, f'Methylated: 0/{total_cpgs} CpGs',
         fontsize=9, color='black')

# --- Cancer tissue plot ---
for pos in cancer_unmethylated:
    ax2.plot([pos, pos], [0, 1], color='gray', linewidth=0.8)
    ax2.scatter(pos, 1, s=80, facecolors='white',
                edgecolors='black', linewidth=1.5, zorder=3)

for pos in cancer_methylated:
    ax2.plot([pos, pos], [0, 1], color='gray', linewidth=0.8)
    ax2.scatter(pos, 1, s=80, facecolors='black',
                edgecolors='black', linewidth=1.5, zorder=3)

ax2.axhline(y=0, color='black', linewidth=2)
ax2.set_xlim(0, len(sequence))
ax2.set_ylim(-0.5, 1.8)
ax2.set_yticks([])
ax2.set_title('Colorectal Cancer (HCT116) — MLH1 Promoter CpG Methylation',
              fontsize=12, fontweight='bold')
ax2.set_xlabel('Position (bp)')
ax2.text(10, 1.4,
         f'Methylated: {len(cancer_methylated)}/{total_cpgs} CpGs',
         fontsize=9, color='black')

# Legend
open_circle = mpatches.Patch(facecolor='white',
                              edgecolor='black', label='Unmethylated CpG')
filled_circle = mpatches.Patch(facecolor='black',
                                edgecolor='black', label='Methylated CpG')
fig.legend(handles=[open_circle, filled_circle],
           loc='lower right', fontsize=10)

plt.suptitle('MLH1 Promoter CpG Methylation: Normal vs Colorectal Cancer',
             fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('MLH1_methylation_lollipop.png', dpi=150, bbox_inches='tight')
plt.show()
print("Plot saved!")
```

```python
categories = ['Unmethylated CpGs', 'Methylated CpGs']
normal_values = [len(normal_unmethylated), len(normal_methylated)]
cancer_values = [len(cancer_unmethylated), len(cancer_methylated)]

x = range(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(7, 5))
bars1 = ax.bar([i - width/2 for i in x], normal_values,
               width, label='Normal', color='#2ecc71', edgecolor='black')
bars2 = ax.bar([i + width/2 for i in x], cancer_values,
               width, label='Cancer (HCT116)', color='#e74c3c',
               edgecolor='black')

ax.set_xticks(list(x))
ax.set_xticklabels(categories)
ax.set_ylabel('Number of CpG Sites')
ax.set_title('MLH1 Promoter CpG Methylation Status\nNormal vs Colorectal Cancer')
ax.legend()

for bar in bars1 + bars2:
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 0.3,
            str(int(bar.get_height())),
            ha='center', fontsize=10)

plt.tight_layout()
plt.savefig('MLH1_methylation_bargraph.png', dpi=150)
plt.show()
```
